```python
# find possible outliers with the help of this function
def find_outliers(column):

    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = data[(data[column] < lower_limit) | (data[column] > upper_limit)]

    return outliers, lower_limit, upper_limit
```

```python
# checking high or low yield value
yield_outliers, lower, upper = find_outliers(
    'Yield_Tonnes_Ha'
)

print("Lower limit: ", lower)
print("Upper limit: ", upper)
print("Possible yield outliers: ", len(yield_outliers))
```

<b> What insights can be derived from the observed seasonal differences? </b>

```python
# summarize the most important agricultural indicators
# for each season.

season_summary = data.groupby('Season').agg({
    'Yield_Tonnes_Ha': 'mean',
    'Production_Tonnes': 'mean',
    'Rainfall_mm': 'mean',
    'Avg_Temperature_C': 'mean',
    'Soil_Moisture_pct': 'mean',
    'Water_Used_m3': 'mean',
    'Water_Efficiency_t_per_1000m3': 'mean',
    'Revenue_INR': 'mean',
    'Total_Cost_INR': 'mean',
    'Profit_INR': 'mean',
    'Disease_Pest_Risk_pct': 'mean'
}).round(2)

season_summary
```

```python
# Finding the best and worst seasons for important indicators

best_yield = season_summary['Yield_Tonnes_Ha'].idxmax()
worst_yield = season_summary['Yield_Tonnes_Ha'].idxmin()

best_profit = season_summary['Profit_INR'].idxmax()
worst_profit = season_summary['Profit_INR'].idxmin()

best_water_efficiency = season_summary[
    'Water_Efficiency_t_per_1000m3'
].idxmax()

highest_water_use = season_summary[
    'Water_Used_m3'
].idxmax()

highest_revenue = season_summary[
    'Revenue_INR'
].idxmax()

print("Highest average yield:", best_yield)
print("Lowest average yield:", worst_yield)

print("Highest average profit:", best_profit)
print("Lowest average profit:", worst_profit)

print("Best water efficiency:", best_water_efficiency)
print("Highest water usage:", highest_water_use)

print("Highest average revenue:", highest_revenue)
```

```python
# Measuring how much the best-performing season differs from the lowest-performing season.

highest_yield_value = season_summary.loc[
    best_yield, 'Yield_Tonnes_Ha'
]

lowest_yield_value = season_summary.loc[
    worst_yield, 'Yield_Tonnes_Ha'
]

yield_difference_percent = (
    (highest_yield_value - lowest_yield_value)
    / lowest_yield_value
) * 100

print(
    "Yield difference between highest and lowest season:",
    round(yield_difference_percent, 2),
    "%"
)
```

```python
# Creating a simple interpretation from our results

print("========== KEY SEASONAL INSIGHTS ==========\n")

print(
    f"1. {best_yield} has the highest average crop yield, "
    f"while {worst_yield} has the lowest."
)

print(
    f"2. {best_profit} has the highest average profit, "
    f"while {worst_profit} has the lowest."
)

print(
    f"3. {best_water_efficiency} shows the highest "
    f"average water efficiency."
)

print(
    f"4. {highest_water_use} has the highest average "
    f"water consumption."
)

print(
    f"5. {highest_revenue} generates the highest "
    f"average revenue."
)

print(
    f"6. The difference between the highest and lowest "
    f"average yield is approximately "
    f"{yield_difference_percent:.2f}%."
)

print("\n============================================")
```

```python
# Creating recommendations based on the observed indicators

recommendations = []

for season in season_summary.index:

    profit = season_summary.loc[
        season, 'Profit_INR'
    ]

    water_efficiency = season_summary.loc[
        season, 'Water_Efficiency_t_per_1000m3'
    ]

    risk = season_summary.loc[
        season, 'Disease_Pest_Risk_pct'
    ]

    if profit > 0:
        economic_action = "Evaluate as economically favorable"
    else:
        economic_action = "Review costs and market conditions carefully"

    if water_efficiency < season_summary[
        'Water_Efficiency_t_per_1000m3'
    ].mean():
        water_action = "Focus on irrigation efficiency"
    else:
        water_action = "Maintain efficient water management"

    recommendations.append([
        season,
        economic_action,
        water_action
    ])

recommendation_table = pd.DataFrame(
    recommendations,
    columns=[
        'Season',
        'Economic Planning',
        'Water Planning'
    ]
)

recommendation_table
```
